```python
df = pd.read_excel('/content/drive/MyDrive/Datasets/Customer Call List.xlsx')
df
```

```python
#removing duplicates
df = df.drop_duplicates()
df
```

```python
#removing unnecessary data
df = df.drop(columns='Not_Useful_Column')
df
```

```python
#cleaning Last_Name column
df['Last_Name'] = df['Last_Name'].str.lstrip('...')
df['Last_Name'] = df['Last_Name'].str.lstrip('/')
df['Last_Name'] = df['Last_Name'].str.rstrip('_')
#df['Last_Name'] = df['Last_Name'].str.strip('123._/')
df
```

```python
df
```

```python
#cleaning Phone_Number column
df['Phone_Number'] = df['Phone_Number'].str.replace(r'[^a-zA-Z0-9]','',regex = True)
```

```python
df['Phone_Number']
```

```python
df['Phone_Number'] = df['Phone_Number'].apply(lambda x: str(x))
df['Phone_Number'] = df['Phone_Number'].apply(lambda x: x[0:3] + '-' + x[3:6] + '-' + x[6:10])
df['Phone_Number']
```

```python
df['Phone_Number'] = df['Phone_Number'].str.replace('nan--', '', regex=False)
df['Phone_Number'] = df['Phone_Number'].str.replace('Na--', '', regex=False)
df['Phone_Number']
```

```python
df
```

```python
#handling the address
df[['Street_Address', 'State', 'Zip_Code']] = df['Address'].str.split(',', expand=True)
df
```

```python
#standardizing Paying_Customer column
df['Paying Customer'] = df['Paying Customer'].str.replace('Yes', 'Y')
df['Paying Customer'] = df['Paying Customer'].str.replace('No', 'N')
df['Paying Customer']
```

```python
df
```

```python
#standardizing Do_Not_Contact column
df['Do_Not_Contact'] = df['Do_Not_Contact'].str.replace('Yes', 'Y')
df['Do_Not_Contact'] = df['Do_Not_Contact'].str.replace('No', 'N')
df['Do_Not_Contact']
```

```python
df
```

```python
#getting rid of N/a values
df = df.replace('N/a', '')
df
```

```python
#getting rid of NaN values
df = df.fillna('')
```

```python
df
```

```python
#trimming down to only customers that can be contacted
for x in df.index:
  if df.loc[x, 'Do_Not_Contact'] == 'Y':
    df.drop(x, inplace=True)
df
```

```python
for x in df.index:
  if df.loc[x, 'Phone_Number'] == '':
    df.drop(x, inplace=True)
df
```

```python
df = df.reset_index(drop=True)
df
```
